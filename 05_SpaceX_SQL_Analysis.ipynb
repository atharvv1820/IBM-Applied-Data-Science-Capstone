```python
import pandas as pd
import sqlite3
from IPython.display import display

df = pd.read_csv("/content/dataset_part_2.csv")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
display(df.head())

conn = sqlite3.connect(":memory:")
df.to_sql("spacex", conn, index=False, if_exists="replace")

print("SQL table 'spacex' created successfully.")
```

```python
query = """
SELECT *
FROM spacex
LIMIT 10;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT LaunchSite, COUNT(*) AS TotalLaunches
FROM spacex
GROUP BY LaunchSite
ORDER BY TotalLaunches DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT MAX(PayloadMass) AS MaximumPayloadMass
FROM spacex;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Outcome, COUNT(*) AS Total
FROM spacex
GROUP BY Outcome
ORDER BY Total DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    Orbit,
    COUNT(*) AS LaunchCount,
    ROUND(AVG(PayloadMass), 2) AS AveragePayloadMass
FROM spacex
WHERE PayloadMass IS NOT NULL
GROUP BY Orbit
ORDER BY AveragePayloadMass DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Class, COUNT(*) AS Total
FROM spacex
GROUP BY Class
ORDER BY Class;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT FlightNumber, PayloadMass, Outcome
FROM spacex
WHERE Outcome = 'True ASDS'
ORDER BY FlightNumber;
"""

display(pd.read_sql_query(query, conn))
```
