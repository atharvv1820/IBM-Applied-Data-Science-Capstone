```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from io import StringIO

print("Libraries imported successfully")
```

```python
url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"

headers = {
    "User-Agent": "Mozilla/5.0 (compatible; CapstoneProject/1.0)"
}

response = requests.get(url, headers=headers, timeout=30)

print("HTTP status:", response.status_code)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

print("Page title:", soup.title.get_text(strip=True))
```

```python
tables = pd.read_html(StringIO(response.text))

print("Number of tables found:", len(tables))

for i, table in enumerate(tables):
    print(f"\nTable {i}: {table.shape}")
    print(table.columns.tolist()[:8])
```

```python
launch_table = None

for table in tables:
    columns = " ".join(str(col) for col in table.columns).lower()

    if "flight" in columns and "date" in columns:
        launch_table = table.copy()
        break

if launch_table is None:
    raise ValueError(
        "Could not identify the launch table. "
        "Inspect the table list from Cell 3."
    )

launch_table.columns = [
    " ".join(str(col) for col in column).strip()
    if isinstance(column, tuple)
    else str(column).strip()
    for column in launch_table.columns
]

launch_table = launch_table.dropna(how="all")

print("Extracted table shape:", launch_table.shape)
display(launch_table.head())
```

```python
output_file = "spacex_web_scraped_from_web.csv"

launch_table.to_csv(output_file, index=False)

print("Saved:", output_file)
print("Rows:", len(launch_table))
print("Columns:", len(launch_table.columns))
```
