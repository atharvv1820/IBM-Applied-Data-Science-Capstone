```python
import pandas as pd

geo_df = pd.read_csv("/content/spacex_launch_geo.csv")

print("Dataset shape:", geo_df.shape)
print("Columns:", geo_df.columns.tolist())
display(geo_df.head())
```

```python
for col in geo_df.columns:
    print(col, ":", geo_df[col].dtype)
```

```python
print("Available columns:")
print(geo_df.columns.tolist())

print("\nFirst 5 rows:")
display(geo_df.head())
```

```python
# Clean column names
geo_df.columns = geo_df.columns.str.strip()

# Automatically find latitude and longitude columns
LAT_COL = next(
    (col for col in geo_df.columns
     if "latitude" in col.lower() or col.lower() in ["lat", "latitudes"]),
    None
)

LON_COL = next(
    (col for col in geo_df.columns
     if "longitude" in col.lower() or col.lower() in ["lon", "lng", "long"]),
    None
)

print("Latitude column:", LAT_COL)
print("Longitude column:", LON_COL)

if LAT_COL is None or LON_COL is None:
    raise ValueError(
        "Coordinate columns not detected. Check the column names printed in Step 1."
    )

# Convert coordinates to numeric
geo_df[LAT_COL] = pd.to_numeric(geo_df[LAT_COL], errors="coerce")
geo_df[LON_COL] = pd.to_numeric(geo_df[LON_COL], errors="coerce")

# Remove missing or invalid coordinates
geo_df = geo_df.dropna(subset=[LAT_COL, LON_COL])

# Keep valid geographic coordinates only
geo_df = geo_df[
    geo_df[LAT_COL].between(-90, 90) &
    geo_df[LON_COL].between(-180, 180)
].copy()

print("Valid coordinate rows:", len(geo_df))
display(geo_df.head())
```

```python
!pip -q install folium

import folium

# Center the map around the launch sites
center_lat = geo_df[LAT_COL].mean()
center_lon = geo_df[LON_COL].mean()

launch_map = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=4
)

# Add a marker for every launch
for _, row in geo_df.iterrows():
    folium.Marker(
        location=[row[LAT_COL], row[LON_COL]],
        popup=(
            f"Flight Number: {row['Flight Number']}<br>"
            f"Launch Site: {row['Launch Site']}<br>"
            f"Date: {row['Date']}<br>"
            f"Landing Outcome: {row['Landing Outcome']}"
        ),
        tooltip=row["Launch Site"]
    ).add_to(launch_map)

launch_map
```

```python
launch_map.save("spacex_launch_map_created.html")

print("Map saved successfully!")
```
